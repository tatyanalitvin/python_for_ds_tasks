### Завдання 1: Створення матриць та векторів
1. Створіть `numpy` вектор `vector` розмірністю 10 з випадковими дійсними значеннями в діапазоні від 0 до 1. Можна скористатись двома альтернативними варіантами:
  1. Генерація випадкових чисел з модулем [`random.uniform`](https://docs.python.org/3/library/random.html).
  2. Створення вектора з допомогою метода `numpy` [np.random.rand](https://numpy.org/doc/stable/reference/random/generated/numpy.random.rand.html)

2. Створіть матрицю `matrix` типу `numpy.array` розмірності 3x3 з наступними значеннями

  $$
  matrix = \begin{bmatrix}
  11 & 14 & 17 \\
  20 & 23 & 26 \\
  29 & 32 & 35
  \end{bmatrix}
  $$

  будь-яким відомими Вам способом.

Виведіть на екран вектор і матрицю, а також їх розміри.


```python
# Генерація випадкових чисел з модулем random.uniform.
vector = np.random.uniform(size=10)

print(f"""
Vector:
{vector}

Type:
{type(vector)}

Shape:
{vector.shape}
"""
)
```

```python
# Створення вектора з допомогою метода numpy np.random.rand
vector = np.random.rand(10, )
print(f"""
Vector:
{vector}

Type:
{type(vector)}

Shape:
{vector.shape}
"""
)
```

```python
# Створіть матрицю matrix типу numpy.array розмірності 3x3 з наступними значеннями
matrix = np.array([
    [11, 14, 17],
    [20, 23, 26],
    [29, 32, 35]
  ])

print(f"""
Vector:
{matrix}

Type:
{type(matrix)}

Shape:
{matrix.shape}
"""
)
```


### Завдання 3: Множення матриць
1. Задані дві матриці `A` і `B`розмірностями 3x2 та 2х3. Виконайте їхнє множення, щоб отримати матрицю 2x2. Виведіть результуючу матрицю `result_matrix` на екран.

2. Заданий вектор `C` розмірністю 2х1. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на вектор `C` і який результат буде. А потім виконайте множення і виведіть результат на екран.

3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

```python
# Задані дві матриці A і Bрозмірностями 3x2 та 2х3.
# Виконайте їхнє множення, щоб отримати матрицю 2x2.
# Виведіть результуючу матрицю result_matrix на екран.
result_matrix = np.dot(B, A)
display(Matrix(A), Math('\\times'), Matrix(B), Math('='), Matrix(result_matrix))
```

```python
# Заданий вектор C розмірністю 2х1.
# Подумайте спочатку, яку розмірність Ви отрмаєте,
# коли домножите result_matrix на вектор C і який результат буде.
# А потім виконайте множення і виведіть результат на екран.
display(Matrix(result_matrix), Math('\\times'), Matrix(C), Math('='), Matrix(np.dot(result_matrix, C)))
```

```python
# Задана матриця D розмірністю 2х2.
# Подумайте спочатку, яку розмірність Ви отрмаєте,
# коли домножите result_matrix на матрицю D і який результат буде.
# А потім виконайте множення і виведіть результат на екран.
```

```python
display(Matrix(result_matrix), Math('\\times'), Matrix(D), Math('='), Matrix(np.dot(result_matrix, D)))
```

```python
# Транспонуйте result_matrix.
print("Оригінальна result_matrix:")
display(Matrix(result_matrix))

print("\nТранспонована result_matrix:")
display(Matrix(result_matrix.T))
```

```python
# Зробіть транспонування вектора vector і виведіть форму нової структури.
# Чи бачите ви зміну?
print(f"""
Vector:
{vector}
Shape:
{vector.shape}

Vector T:
{vector.T}
Shape:
{vector.T.shape}
"""
)
```

```python
# Використайте операцію numpy.expand_dims над вектором vector з аргументом axis=1.
# Запишіть отриманий результат в змінну column_vector і виведіть на екран.
# Порівняйте з vector.
column_vector = np.expand_dims(vector, axis=1)
print(f"""
Vector:
{vector}
Shape:
{vector.shape}

Vector with expanded dimensions:
{column_vector}
Shape:
{column_vector.shape}
"""
)
```
